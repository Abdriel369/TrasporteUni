```python
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.svm import SVR
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
from sklearn.neighbors import KNeighborsRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.svm import SVR
import pandas as pd

# Escalamos los datos (SVM y regresión lineal lo necesitan; RandomForest no, pero no le afecta)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

modelos = {
    # Tus modelos originales
    "RandomForest": RandomForestRegressor(n_estimators=200, random_state=42),
    "SVR (rbf)": SVR(kernel="rbf", C=100, epsilon=0.5),
    "SVR (linear)": SVR(kernel="linear", C=100, epsilon=0.5),
    "LinearRegression": LinearRegression(),
    # --- Nuevos modelos ---
    # Red Neuronal (Perceptrón Multicapa)
    "Red Neuronal (MLP)": MLPRegressor(
        hidden_layer_sizes=(128, 64), max_iter=500, random_state=42
    ),
    # Modelo de Boosting (Suele superar a RandomForest)
    "GradientBoosting": GradientBoostingRegressor(
        n_estimators=200, learning_rate=0.1, random_state=42
    ),
    # Modelo basado en vecinos cercanos
    "KNN": KNeighborsRegressor(n_neighbors=5),
}
resultados = []

for nombre, modelo in modelos.items():
    modelo.fit(X_train_scaled, y_train)
    y_pred = modelo.predict(X_test_scaled)

    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)

    resultados.append({"Modelo": nombre, "MAE": mae, "RMSE": rmse, "R2": r2})

tabla_resultados = pd.DataFrame(resultados).sort_values("R2", ascending=False)
print(tabla_resultados)
```

## Modelos No Supervisados aplicados a la Demanda de Viajes

**Aclaración importante antes de empezar:** el clustering (K-Means, GMM, etc.) es un algoritmo *no supervisado*: no usa la variable `label` (demanda) para entrenarse, así que no puede "predecir" un número por sí solo como lo hace una regresión.

Lo que sí podemos hacer, y es una práctica común en ciencia de datos, es:

1. **Segmentar** los viajes en grupos (clusters) usando solo las variables predictoras (`X`), sin tocar `label`.
2. **Analizar** si esos grupos, formados sin supervisión, terminan separando bien niveles de demanda alta/media/baja.
3. **Cluster-then-predict**: una vez formado el cluster (sin usar `y`), calculamos el promedio de `label` de cada cluster en el set de entrenamiento y lo usamos como predicción para nuevos datos (asignándolos al cluster más cercano). Es una forma válida de convertir clustering en un predictor simple, y nos sirve para comparar contra tus modelos supervisados.
4. **Feature engineering híbrido**: usar el número de cluster como una variable más dentro de tu mejor modelo supervisado (Gradient Boosting), a ver si mejora el R².

### 1. Encontrar el número óptimo de clusters (Método del Codo + Silhouette)

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Escalamos SOLO con las variables predictoras (X), el clustering nunca ve 'label'
scaler_unsup = StandardScaler()
X_train_scaled_u = scaler_unsup.fit_transform(X_train)
X_test_scaled_u = scaler_unsup.transform(X_test)

inertias = []
silhouettes = []
rango_k = range(2, 11)

for k in rango_k:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels_k = km.fit_predict(X_train_scaled_u)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_train_scaled_u, labels_k))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(list(rango_k), inertias, marker='o')
axes[0].set_title('Método del Codo')
axes[0].set_xlabel('Número de clusters (k)')
axes[0].set_ylabel('Inercia')

axes[1].plot(list(rango_k), silhouettes, marker='o', color='orange')
axes[1].set_title('Silhouette Score')
axes[1].set_xlabel('Número de clusters (k)')
axes[1].set_ylabel('Silhouette')

plt.tight_layout()
plt.show()

print("Silhouette por k:", dict(zip(rango_k, [round(s,3) for s in silhouettes])))
```

### 2. Entrenar K-Means con el k elegido y analizar los clusters vs. la demanda real

```python
# Ajusta este valor según el codo/silhouette que hayas observado arriba
K_OPTIMO = 4

kmeans = KMeans(n_clusters=K_OPTIMO, random_state=42, n_init=10)
cluster_train = kmeans.fit_predict(X_train_scaled_u)
cluster_test = kmeans.predict(X_test_scaled_u)

# Vemos si los clusters (formados SIN ver la demanda) separan bien la demanda real
analisis = pd.DataFrame({'cluster': cluster_train, 'label': y_train.values})
resumen_clusters = analisis.groupby('cluster')['label'].agg(['mean', 'median', 'std', 'count'])
print(resumen_clusters)

plt.figure(figsize=(7, 4))
analisis.boxplot(column='label', by='cluster')
plt.title('Distribución de la demanda por cluster')
plt.suptitle('')
plt.xlabel('Cluster')
plt.ylabel('Demanda (label)')
plt.show()
```

### 3. Visualización con PCA (reducimos a 2D para ver los clusters)

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2, random_state=42)
X_train_pca = pca.fit_transform(X_train_scaled_u)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sc1 = axes[0].scatter(X_train_pca[:, 0], X_train_pca[:, 1], c=cluster_train, cmap='tab10', s=15)
axes[0].set_title('Clusters (K-Means)')
axes[0].set_xlabel('PC1')
axes[0].set_ylabel('PC2')
plt.colorbar(sc1, ax=axes[0], label='Cluster')

sc2 = axes[1].scatter(X_train_pca[:, 0], X_train_pca[:, 1], c=y_train, cmap='viridis', s=15)
axes[1].set_title('Demanda real (label)')
axes[1].set_xlabel('PC1')
axes[1].set_ylabel('PC2')
plt.colorbar(sc2, ax=axes[1], label='Demanda')

plt.tight_layout()
plt.show()

print(f"Varianza explicada por las 2 componentes: {pca.explained_variance_ratio_.sum():.2%}")
```

### 4. "Cluster-then-predict": usar el promedio de demanda por cluster como predicción

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Promedio de demanda por cluster, calculado SOLO con datos de entrenamiento
demanda_promedio_por_cluster = analisis.groupby('cluster')['label'].mean()

# A cada viaje de test le asignamos la demanda promedio de su cluster más cercano
y_pred_cluster = pd.Series(cluster_test).map(demanda_promedio_por_cluster).values

mae_c = mean_absolute_error(y_test, y_pred_cluster)
rmse_c = np.sqrt(mean_squared_error(y_test, y_pred_cluster))
r2_c = r2_score(y_test, y_pred_cluster)

print(f"K-Means (cluster-then-predict) con k={K_OPTIMO}")
print(f"MAE:  {mae_c:.2f}")
print(f"RMSE: {rmse_c:.2f}")
print(f"R²:   {r2_c:.3f}")
```

### 5. Alternativa: Gaussian Mixture Model (clustering probabilístico)

```python
from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(n_components=K_OPTIMO, random_state=42)
gmm_cluster_train = gmm.fit_predict(X_train_scaled_u)
gmm_cluster_test = gmm.predict(X_test_scaled_u)

analisis_gmm = pd.DataFrame({'cluster': gmm_cluster_train, 'label': y_train.values})
demanda_promedio_gmm = analisis_gmm.groupby('cluster')['label'].mean()

y_pred_gmm = pd.Series(gmm_cluster_test).map(demanda_promedio_gmm).values

mae_g = mean_absolute_error(y_test, y_pred_gmm)
rmse_g = np.sqrt(mean_squared_error(y_test, y_pred_gmm))
r2_g = r2_score(y_test, y_pred_gmm)

print(f"Gaussian Mixture (cluster-then-predict) con {K_OPTIMO} componentes")
print(f"MAE:  {mae_g:.2f}")
print(f"RMSE: {rmse_g:.2f}")
print(f"R²:   {r2_g:.3f}")
```

### 6. Enfoque híbrido: usar el cluster como variable extra dentro de tu mejor modelo supervisado

```python
from sklearn.ensemble import GradientBoostingRegressor

# Agregamos el número de cluster como una feature más
X_train_hibrido = X_train.copy()
X_train_hibrido['cluster'] = cluster_train

X_test_hibrido = X_test.copy()
X_test_hibrido['cluster'] = cluster_test

modelo_hibrido = GradientBoostingRegressor(
    n_estimators=300, learning_rate=0.05, max_depth=5, subsample=0.8, random_state=42
)
modelo_hibrido.fit(X_train_hibrido, y_train)
y_pred_hibrido = modelo_hibrido.predict(X_test_hibrido)

mae_h = mean_absolute_error(y_test, y_pred_hibrido)
rmse_h = np.sqrt(mean_squared_error(y_test, y_pred_hibrido))
r2_h = r2_score(y_test, y_pred_hibrido)

print("Gradient Boosting + feature de cluster (K-Means)")
print(f"MAE:  {mae_h:.2f}")
print(f"RMSE: {rmse_h:.2f}")
print(f"R²:   {r2_h:.3f}")
```

### 7. Tabla comparativa: supervisado puro vs. no supervisado vs. híbrido

```python
comparacion_final = pd.DataFrame([
    {'Enfoque': 'K-Means (cluster-then-predict)', 'MAE': mae_c, 'RMSE': rmse_c, 'R2': r2_c},
    {'Enfoque': 'GMM (cluster-then-predict)', 'MAE': mae_g, 'RMSE': rmse_g, 'R2': r2_g},
    {'Enfoque': 'GradientBoosting + cluster (híbrido)', 'MAE': mae_h, 'RMSE': rmse_h, 'R2': r2_h},
]).sort_values('R2', ascending=False)

print(comparacion_final)
```

**Conclusión esperada:** los enfoques puramente no supervisados (K-Means, GMM usados como predictor) casi siempre van a rendir por debajo de tus mejores modelos supervisados (Gradient Boosting, Random Forest), porque están prediciendo con el promedio de un grupo en vez de aprender una función continua. Su valor real está en:
- Descubrir **segmentos naturales** de demanda (horas pico, zonas, etc.) que quizás no habías definido a mano.
- Servir como **variable adicional** para el modelo supervisado (paso 6), lo cual a veces sí mejora el R² un poco.

Si el objetivo final es la mejor predicción posible, tus modelos supervisados (especialmente Gradient Boosting con ~0.76-0.79 de R²) siguen siendo el camino correcto; lo no supervisado aquí es complementario, no un reemplazo.
